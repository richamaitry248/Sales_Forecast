```python
def split_data(state_df):
    # Sort to be 100% sure
    state_df = state_df.sort_values('Date')
    
    # The last 8 weeks are for testing
    train = state_df.iloc[:-8]
    test = state_df.iloc[-8:]
    
    return train, test
```
